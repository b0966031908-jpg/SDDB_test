## 4. AI 的工作
- 輸入：使用者的一句話
- 輸出格式：`動作,品項,數量`
- 動作只能是：訂購、取消、查詢、統計、無法判斷

- 數量的規則：以阿拉伯數字輸出，中文數字要轉換（兩杯→2、三杯→3）；使用者沒說數量時視為 1；取消、查詢、統計、無法判斷一律輸出 0；數量不在 1～3 之間時，AI 照實輸出抽出的數字，由程式負責拒絕
- 無法判斷時輸出：`無法判斷,無,0`
- AI 不可以做的事：不可輸出格式以外的文字或解釋；不可自行扣庫存、算金額或修改資料庫；不可編造菜單以外的品項；不可回答閒聊或與咖啡廳無關的問題；不可判斷數量是否超過上限（由程式判斷）

## 6. 例外情況（至少 3 個）
1. 使用者要菜單以外的品項（如珍珠奶茶）或說閒聊的話：AI 輸出「無法判斷,無,0」，系統回覆目前有哪些品項，資料庫不變。
2. 數量為 0、負數或超過 3（例如「我要五杯拿鐵」）：由程式拒絕並提示數量需在 1～3 之間，庫存與訂單都不變。
3. 庫存不足（例如抹茶歐蕾只剩 3 杯卻訂 3 杯後再有人訂）：拒絕並告知剩餘杯數，庫存不會變成負數。
4. 取消時查不到訂單，或同一筆訂單重複取消：回覆查無訂單，不會重複釋出庫存（已取消的訂單不會再被取消）。
5. 沒輸入姓名卻要訂購或取消：提示請輸入姓名，不做任何處理。
6. 單一學生累計超過 3 杯（所有品項加總）：拒絕訂購；先取消舊訂單釋出額度後即可再訂。

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
DB_PATH = "cafe.db"            # 資料庫檔案名稱

# --- 規則常數（改規則只要改這裡）---
ITEMS = ["美式咖啡", "拿鐵", "抹茶歐蕾"]                  # 允許的品項
ACTIONS = ["訂購", "取消", "查詢", "統計", "無法判斷"]    # 允許的動作
STATUS_ORDERED = "已訂購"
STATUS_CANCELLED = "已取消"
MIN_QTY = 1                    # 單次數量下限
MAX_QTY = 3                    # 單次數量上限
PER_PERSON_LIMIT = 3           # 每人今天最多訂幾杯（所有品項加總）
UNKNOWN_MSG = "無法判斷，目前提供的品項有：" + "、".join(ITEMS)

# 初始資料：(編號, 品項, 單價, 庫存)
INITIAL_MENU = [
    (1, "美式咖啡", 50, 5),
    (2, "拿鐵", 65, 4),
    (3, "抹茶歐蕾", 70, 3),
]


def get_conn():
    """開啟一個資料庫連線（用完一定要 close）"""
    return sqlite3.connect(DB_PATH)


def init_db():
    """重建資料表並放入初始資料"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        # 先刪舊表（orders 參照 menu，所以先刪 orders）
        cur.execute("DROP TABLE IF EXISTS orders")
        cur.execute("DROP TABLE IF EXISTS menu")

        # 菜單表：品項、單價、目前庫存
        cur.execute("""
            CREATE TABLE menu (
                id    INTEGER PRIMARY KEY,
                name  TEXT NOT NULL,
                price INTEGER NOT NULL,
                stock INTEGER NOT NULL
            )
        """)
        # 訂單表：誰、訂了哪個品項、幾杯、目前狀態
        cur.execute("""
            CREATE TABLE orders (
                id      INTEGER PRIMARY KEY AUTOINCREMENT,
                student TEXT NOT NULL,
                item_id INTEGER NOT NULL,
                qty     INTEGER NOT NULL,
                status  TEXT NOT NULL,
                FOREIGN KEY (item_id) REFERENCES menu(id)
            )
        """)
        # 放入初始菜單（用 ? 參數，不用 f-string 組 SQL）
        cur.executemany(
            "INSERT INTO menu (id, name, price, stock) VALUES (?, ?, ?, ?)",
            INITIAL_MENU,
        )
        conn.commit()
    finally:
        conn.close()


init_db()
print("✅ 資料庫已建立並放入初始資料")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
def show_tables():
    """顯示 menu 與 orders 兩張資料表的目前內容"""
    conn = get_conn()
    try:
        menu_df = pd.read_sql_query("SELECT * FROM menu", conn)
        orders_df = pd.read_sql_query("SELECT * FROM orders", conn)
    finally:
        conn.close()
    print("📋 菜單 menu")
    display(menu_df)
    print("📋 訂單 orders")
    display(orders_df)


show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
SYSTEM_PROMPT = """你是咖啡廳點餐系統的語意解析器。
請把使用者的一句話解析成一行文字，格式固定為：動作,品項,數量

規則：
1. 動作只能是：訂購、取消、查詢、統計、無法判斷。
2. 品項只能是：美式咖啡、拿鐵、抹茶歐蕾。動作是統計或無法判斷時，品項填「無」。
3. 數量用阿拉伯數字（兩杯=2、三杯=3）；訂購時使用者沒說數量就填 1；
   使用者說幾杯就照實填，即使超過 3 也照填；取消、查詢、統計、無法判斷一律填 0。
4. 問今天各品項賣得怎樣、銷售狀況、營收，動作是統計。
5. 菜單以外的品項、閒聊或與咖啡廳無關的話，一律輸出：無法判斷,無,0
6. 只輸出那一行，不要任何解釋、標點或其他文字。

範例：
我要訂兩杯拿鐵 -> 訂購,拿鐵,2
幫我取消抹茶歐蕾 -> 取消,抹茶歐蕾,0
美式咖啡還有幾杯 -> 查詢,美式咖啡,0
今天各品項賣得怎樣 -> 統計,無,0
我要一杯珍珠奶茶 -> 無法判斷,無,0
"""

MAX_RETRY = 3          # 遇到 429 最多重試 3 次
ai_cache = {}          # 暫存：同一句話只呼叫一次 AI


def is_rate_limit(err):
    """判斷錯誤是不是「超過使用量（429）」"""
    msg = str(err).lower()
    return getattr(err, "status_code", None) == 429 or "429" in msg or "rate limit" in msg


def call_ai(text):
    """呼叫 Groq，回傳 AI 原始回覆文字；失敗回傳空字串"""
    key = text.strip()
    if key in ai_cache:                 # 問過了就直接用暫存，不再呼叫
        return ai_cache[key]

    raw = ""
    for attempt in range(MAX_RETRY + 1):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": key},
                ],
            )
            raw = (resp.choices[0].message.content or "").strip()
            ai_cache[key] = raw         # 成功才存入暫存
            break
        except Exception as e:
            if is_rate_limit(e) and attempt < MAX_RETRY:
                print(f"⏳ 超過使用量，等待 30 秒後重試（第 {attempt + 1} 次）")
                time.sleep(30)
                continue
            print("⚠️ AI 呼叫失敗：", type(e).__name__)
            break
        finally:
            time.sleep(2)               # 每次呼叫後休息 2 秒
    return raw


def parse_reply(raw):
    """從 AI 回覆中，找出允許清單裡的動作、品項、數量"""
    # 動作：找出現位置最前面的允許動作，找不到就是「未知」
    action = "未知"
    best = None
    for a in ACTIONS:
        pos = raw.find(a)
        if pos != -1 and (best is None or pos < best):
            best, action = pos, a

    # 品項：同樣只接受允許清單裡的值
    item = "未知"
    best = None
    for it in ITEMS:
        pos = raw.find(it)
        if pos != -1 and (best is None or pos < best):
            best, item = pos, it

    # 數量：用正規表示式找第一個數字，找不到就當作 1
    m = re.search(r"\d+", raw)
    qty = int(m.group()) if m else 1

    # 只有「訂購」才需要數量；其他動作數量固定 0，品項只有訂購/取消/查詢需要
    if action != "訂購":
        qty = 0
    if action not in ("訂購", "取消", "查詢"):
        item = "無"
    return action, item, qty


def interpret(text):
    """完整解讀一句話：回傳 {raw, action, item, qty}"""
    raw = call_ai(text)
    action, item, qty = parse_reply(raw)
    return {"raw": raw, "action": action, "item": item, "qty": qty}


print("✅ AI 解讀函式已定義")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
ai_tests = [
    ("我要訂兩杯拿鐵",           "訂購",     "拿鐵",     2),   # 正常
    ("我想要一杯美式咖啡",       "訂購",     "美式咖啡", 1),   # 正常
    ("幫我取消抹茶歐蕾",         "取消",     "抹茶歐蕾", 0),   # 正常
    ("美式咖啡還有幾杯？",       "查詢",     "美式咖啡", 0),   # 正常
    ("今天各品項賣得怎樣？",     "統計",     "無",       0),   # 正常
    ("來杯抹茶歐蕾",             "訂購",     "抹茶歐蕾", 1),   # 邊界：沒說數量
    ("我要五杯拿鐵",             "訂購",     "拿鐵",     5),   # 邊界：超過上限（由程式拒絕）
    ("我要一杯珍珠奶茶",         "無法判斷", "無",       0),   # 錯誤：菜單外品項
    ("今天天氣好嗎",             "無法判斷", "無",       0),   # 錯誤：閒聊
    ("給我三杯抹茶歐蕾啦",       "訂購",     "抹茶歐蕾", 3),   # 口語
]


def match(actual, expected):
    """expected 是 "*" 就一律通過，否則要相等"""
    return expected == "*" or actual == expected


passed = 0
for i, (text, exp_action, exp_item, exp_qty) in enumerate(ai_tests, start=1):
    info = interpret(text)
    ok = (match(info["action"], exp_action)
          and match(info["item"], exp_item)
          and match(info["qty"], exp_qty))
    passed += ok
    print(f"[{i}] {text}")
    print(f"    AI 原始回覆：{info['raw']!r}")
    print(f"    系統解讀：動作={info['action']}、品項={info['item']}、數量={info['qty']}")
    print(f"    預期：動作={exp_action}、品項={exp_item}、數量={exp_qty} → {'✅ 通過' if ok else '❌ 未通過'}")

print(f"\n🎯 AI 解讀測試：通過 {passed} / {len(ai_tests)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================

def find_item(cur, item):
    """用品項名稱查菜單，回傳 (id, price, stock)；找不到回傳 None"""
    cur.execute("SELECT id, price, stock FROM menu WHERE name = ?", (item,))
    return cur.fetchone()


def order_item(student, item, qty):
    """訂購：檢查姓名、數量、庫存、每人上限，成功才寫入"""
    if not student:
        return "請輸入姓名"
    if qty < MIN_QTY or qty > MAX_QTY:
        return f"數量需在 {MIN_QTY}～{MAX_QTY} 杯之間"

    conn = get_conn()
    try:
        cur = conn.cursor()
        row = find_item(cur, item)
        if row is None:                       # fetchone 之後先檢查是否為 None
            return UNKNOWN_MSG
        item_id, price, stock = row

        if stock < qty:
            return f"庫存不足，目前剩餘 {stock} 杯"

        # 這位學生目前有效訂單（已訂購）的總杯數，已取消的不計
        cur.execute(
            "SELECT COALESCE(SUM(qty), 0) FROM orders WHERE student = ? AND status = ?",
            (student, STATUS_ORDERED),
        )
        r = cur.fetchone()
        already = r[0] if r is not None else 0
        if already + qty > PER_PERSON_LIMIT:
            return f"每人今日最多訂 {PER_PERSON_LIMIT} 杯，您已訂 {already} 杯"

        # 通過所有檢查：新增訂單並扣庫存
        cur.execute(
            "INSERT INTO orders (student, item_id, qty, status) VALUES (?, ?, ?, ?)",
            (student, item_id, qty, STATUS_ORDERED),
        )
        cur.execute("UPDATE menu SET stock = stock - ? WHERE id = ?", (qty, item_id))
        conn.commit()
        return f"訂購成功：{student} {item} × {qty} 杯，共 {price * qty} 元"
    finally:
        conn.close()


def cancel_item(student, item):
    """取消：把該學生該品項「最新一筆」已訂購的訂單改成已取消，並釋出庫存"""
    if not student:
        return "請輸入姓名"

    conn = get_conn()
    try:
        cur = conn.cursor()
        row = find_item(cur, item)
        if row is None:
            return UNKNOWN_MSG
        item_id = row[0]

        # id 最大代表最新；已取消的訂單不會被選到，所以不會重複釋出庫存
        cur.execute(
            "SELECT id, qty FROM orders "
            "WHERE student = ? AND item_id = ? AND status = ? "
            "ORDER BY id DESC LIMIT 1",
            (student, item_id, STATUS_ORDERED),
        )
        order = cur.fetchone()
        if order is None:
            return f"查無 {student} 的 {item} 訂單"
        order_id, qty = order

        cur.execute("UPDATE orders SET status = ? WHERE id = ?", (STATUS_CANCELLED, order_id))
        cur.execute("UPDATE menu SET stock = stock + ? WHERE id = ?", (qty, item_id))
        conn.commit()
        return f"已取消：{student} {item} × {qty} 杯"
    finally:
        conn.close()


def query_item(item):
    """查詢：回覆該品項剩餘杯數（唯讀，不需要姓名）"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        row = find_item(cur, item)
        if row is None:
            return UNKNOWN_MSG
        return f"{item} 剩餘 {row[2]} 杯"
    finally:
        conn.close()


def sales_stats():
    """統計：menu LEFT JOIN orders + GROUP BY，沒人訂的品項也會列出（已售 0）"""
    sql = """
        SELECT m.name  AS 品項,
               m.price AS 單價,
               m.stock + COALESCE(SUM(o.qty), 0) AS 今日庫存,
               COALESCE(SUM(o.qty), 0)           AS 已售杯數,
               m.stock                           AS 剩餘杯數,
               COALESCE(SUM(o.qty), 0) * m.price AS 營收
        FROM menu AS m
        LEFT JOIN orders AS o
               ON o.item_id = m.id AND o.status = ?
        GROUP BY m.id, m.name, m.price, m.stock
        ORDER BY m.id
    """
    conn = get_conn()
    try:
        return pd.read_sql_query(sql, conn, params=(STATUS_ORDERED,))
    finally:
        conn.close()


print("✅ 功能函式已定義")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
def handle_request(text, student):
    """主流程：回傳 (系統回應文字, 表格 DataFrame)；沒有表格時回傳空表"""
    student = (student or "").strip()
    empty = pd.DataFrame()
    info = interpret(text)
    action, item, qty = info["action"], info["item"], info["qty"]

    if action in ("訂購", "取消", "查詢"):
        if item not in ITEMS:                 # 品項不在允許清單 → 回覆有哪些品項
            return UNKNOWN_MSG, empty
        if action == "訂購":
            return order_item(student, item, qty), empty
        if action == "取消":
            return cancel_item(student, item), empty
        return query_item(item), empty
    if action == "統計":
        return "今日銷售統計如下", sales_stats()
    return UNKNOWN_MSG, empty                 # 無法判斷或未知


# 先重建資料庫，確保流程測試從初始資料開始
init_db()

# (姓名, 使用者說的話, 回應中應該出現的關鍵字)
flow_tests = [
    ("小明", "我要兩杯拿鐵",        "訂購成功"),       # 拿鐵 4→2
    ("小明", "我要兩杯美式咖啡",    "最多訂 3 杯"),    # 已訂 2 杯，再 2 杯超過上限
    ("小明", "我要一杯美式咖啡",    "訂購成功"),       # 小明累計 3 杯
    ("小華", "我要五杯拿鐵",        "1～3"),           # 數量超出範圍
    ("小華", "我要三杯抹茶歐蕾",    "訂購成功"),       # 抹茶 3→0
    ("小美", "我要一杯抹茶歐蕾",    "庫存不足"),       # 抹茶已賣完
    ("小華", "取消抹茶歐蕾",        "已取消"),         # 抹茶 0→3
    ("小美", "取消拿鐵",            "查無"),           # 小美沒訂過
    ("",     "今天各品項賣得怎樣？", "銷售統計"),       # 統計不需要姓名
    ("",     "拿鐵還有幾杯？",      "剩餘 2 杯"),      # 查詢不需要姓名
]

flow_passed = 0
for i, (name, text, keyword) in enumerate(flow_tests, start=1):
    reply, table = handle_request(text, name)
    ok = keyword in reply
    flow_passed += ok
    print(f"[步驟 {i}] 姓名={name or '（不需姓名）'}｜說：{text}")
    print(f"    系統回應：{reply} → {'✅ 通過' if ok else '❌ 未通過'}")
    if not table.empty:
        display(table)

print(f"\n🎯 流程測試：通過 {flow_passed} / {len(flow_tests)} 步")
print("\n📋 流程測試結束後的資料表：")
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：if IN_COLAB: demo.launch(share=True)
# ============================================================
init_db()

def web_handler(text, student):
    """網頁按下送出時呼叫：回傳 (回應文字, 表格)"""
    if not text or not text.strip():
        return "請輸入您的需求", pd.DataFrame()
    return handle_request(text, student)


with gr.Blocks(title="校園咖啡廳預約取餐系統") as demo:
    gr.Markdown("## ☕ 校園咖啡廳預約取餐系統\n可以說：「我要兩杯拿鐵」、「取消抹茶歐蕾」、「美式咖啡還有幾杯」、「今天各品項賣得怎樣」")
    text_in = gr.Textbox(label="您的需求", placeholder="例如：我要兩杯拿鐵")
    name_in = gr.Textbox(label="姓名（訂購與取消需要，查詢不需要）")
    btn = gr.Button("送出")
    reply_out = gr.Textbox(label="系統回應")
    table_out = gr.Dataframe(label="統計表（統計查詢時顯示）")
    btn.click(web_handler, inputs=[text_in, name_in], outputs=[reply_out, table_out])

if IN_COLAB: demo.launch(share=True)
````
